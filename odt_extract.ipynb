import zipfile
from pathlib import Path
root = Path('c:/Users/flori/Documents/SIMONS-HOMEPAGE')
files = list(root.glob('*.odt'))
print('Found', len(files), 'ODT files')
for odt in files:
    with zipfile.ZipFile(odt, 'r') as z:
        with z.open('content.xml') as f:
            data = f.read().decode('utf-8')
    print('---', odt.name, '---')
    print(data[:2000])
